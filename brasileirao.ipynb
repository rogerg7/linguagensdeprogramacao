```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('brasileirao.xlsx')
```

```python
df.head()
```

Passo 2: Calcular o saldo de gols(SG)

Adicionar uma nova coluna chamada "Saldo de Gols" que será a diferença de Gols Pro(GP) e Gols Contra(GC).

```python
df['Saldo de Gols'] = df['GP'] - df['GC']

df.head()
```

```python
df['P'] = df['V'] * 3 + df['E'] * 1

df
```

```python
df
```

```python
df = df.drop(columns="P")
```

```python
df
```

Filtrar Cluber com mais de 40 Pontos

```python
df
```

```python
clubes_maior_40_pontos = df[df['P '] > 40]
clubes_maior_40_pontos
```

**Passo 5: Visualizar Dados**

Criar um gráfico de barras para visualizar a pontuação dos clubes

```python
df.plot(x='Clube', y='P ', kind='bar', title='Pontuação dos Clubes no Brasileirão 2024')
```
