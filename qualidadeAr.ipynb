Parte 3 — Análise por **cidade**

1. Qual cidade apresenta maior média de PM2.5? São Paulo
2. Qual cidade apresenta maior média de NO2? São Paulo
3. Existe alguma cidade com níveis consistentemente mais altos de poluentes? Sim

```python
df['cidade'].unique()
```

```python
colunas = ['pm25', 'pm10', 'no2', 'o3']
for col in colunas:
    df[col] = pd.to_numeric(df[col], errors='coerce')

resultado = df.groupby('cidade')[colunas].mean()
resultado
```

**Parte 4 — Análise do índice de qualidade do ar**:

1. Qual cidade apresenta pior qualidade média do ar? São Paulo
2. Qual cidade apresenta melhor qualidade do ar? Salvador

```python
df['indice_qualidade_ar'] = pd.to_numeric(df['indice_qualidade_ar'], errors='coerce')
resultado = df.groupby('cidade')['indice_qualidade_ar'].mean().sort_values(ascending=False)
resultado
```

**Parte 7 — Análise temporal**

1. Em quais meses a qualidade do ar tende a ser pior?Junho e Julho
2. Existe algum padrão sazonal?A qualidade do ar piora no outono/inverno e melhora no verão

```python
df['data'] = pd.to_datetime(df['data'], errors='coerce')
df['mes'] = df['data'].dt.month
df['indice_qualidade_ar'] = pd.to_numeric(df['indice_qualidade_ar'], errors='coerce')
resultado = df.groupby('mes')['indice_qualidade_ar'].mean()
resultado
```

**Parte 8 — Correlação entre variáveis**

1. Existe correlação entre temperatura e poluição?Temperatura × PM2.5: -0,21
Temperatura × NO2: -0,19
Temperatura × O3: +0,13
Temperatura × Índice geral: -0,12
2. Quais poluentes parecem mais relacionados entre si?PM10 × NO2: 0,20
PM2.5 × PM10: 0,18
O3 com os demais: praticamente zero

```python
colunas = ['temperatura', 'pm25', 'pm10', 'no2', 'o3']
for col in colunas:
    df[col] = pd.to_numeric(df[col], errors='coerce')

resultado = df[colunas].corr()
resultado
```
